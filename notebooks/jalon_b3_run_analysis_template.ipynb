# Jalon B3 - Run Analysis Template

Template notebook pour charger un run B3 et analyser les artefacts (summary, CSV, attribution map).

```python
from pathlib import Path
import pandas as pd
from boardGPT.interpretability import load_run_artifacts
```

```python
# Remplacer par ton run
run_dir = Path('out_interp/jalon_b3')
runs = sorted([p for p in run_dir.glob('*') if p.is_dir()])
if not runs:
    raise RuntimeError('Aucun run trouve. Lance d abord: python tools/interpretability_cli.py smoke')
active_run = runs[-1]
active_run
```

```python
loaded = load_run_artifacts(str(active_run))
print('config keys:', loaded['config'].keys())
print('summary keys:', loaded['summary'].keys())
print('tensor keys:', loaded['tensors'].keys())
```

```python
metrics_layer_csv = active_run / 'metrics_by_layer.csv'
metrics_overall_csv = active_run / 'metrics_overall.csv'
top_csv = active_run / 'top_contributors.csv'

df_layer = pd.read_csv(metrics_layer_csv)
df_overall = pd.read_csv(metrics_overall_csv)
df_top = pd.read_csv(top_csv)

display(df_overall)
display(df_layer)
display(df_top.head(10))
```

```python
attr_map = loaded['tensors'].get('attribution_map')
if attr_map is None:
    print('No attribution_map tensor found.')
else:
    print('attribution_map shape:', tuple(attr_map.shape))
    # quick textual view
    print(attr_map)
```

```python
report_md = active_run / 'report.md'
print(report_md.read_text(encoding='utf-8')[:1200])
```
