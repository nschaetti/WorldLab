# Jalon B3 - Run From Notebook

Notebook pour lancer un run B3 de bout en bout depuis Python, puis inspecter les artefacts generes.

```python
from pathlib import Path
from boardGPT.interpretability import run_b3_demo_pipeline, load_run_artifacts
```

```python
result = run_b3_demo_pipeline(
    out_root='out_interp/jalon_b3',
    run_name='notebook_run',
    seed=42,
    layers=(0, 1),
    target_cell=5,
    target_class=2,
    target_cells=(3, 5, 9),
    max_examples=8,
)
result['run_dir']
```

```python
run_dir = Path(result['run_dir'])
sorted([p.name for p in run_dir.iterdir()])
```

```python
loaded = load_run_artifacts(str(run_dir))
loaded['summary']
```

```python
report = (run_dir / 'report.md').read_text(encoding='utf-8')
print(report[:1400])
```
